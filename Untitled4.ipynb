```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

noise_levels = [0.05, 0.20, 0.40]
results = []

# Generate synthetic preference data
for noise in noise_levels:
    for method in ["RLHF", "DPO"]:
        scores = []

        for seed in range(5):
            rng = np.random.default_rng(seed)
            true_preferences = rng.random(1000) > 0.5
            noisy_preferences = true_preferences.copy()

            flips = rng.random(1000) < noise
            noisy_preferences[flips] = ~noisy_preferences[flips]

            accuracy = np.mean(
                noisy_preferences == true_preferences
            )

            # Simple illustrative proxy scores
            score = accuracy + rng.normal(0, 0.01)
            scores.append(score)

        results.append({
            "Noise": noise,
            "Method": method,
            "Average Score": np.mean(scores)
        })

df = pd.DataFrame(results)

print("RLHF vs DPO Under Annotator Noise")
print(df.round(3).to_string(index=False))

# Plot comparison
for method in ["RLHF", "DPO"]:
    data = df[df["Method"] == method]
    plt.plot(
        data["Noise"] * 100,
        data["Average Score"],
        marker="o",
        label=method
    )

plt.xlabel("Annotator Noise (%)")
plt.ylabel("Average Score")
plt.title("RLHF vs DPO Under Annotator Noise")
plt.legend()
plt.grid(True)
plt.show()
```

1. Introduction
Aligning language models means shaping their outputs to better match desired behavior. Preference-based methods learn from comparisons between responses rather than requiring a perfect numeric label for every answer. However, preference labels may be inconsistent or noisy. This project asks: as the probability of a flipped preference label rises from 5% to 20% to 40%, how does the measured difference between RLHF and DPO change in a controlled environment?
2. Background
RLHF commonly uses preference comparisons to fit a reward model and then optimizes a policy against that reward, often with a constraint that prevents excessive drift from a reference policy. DPO instead optimizes the policy directly from preferred/rejected pairs relative to a reference policy. Both approaches depend on the quality of preference data. Noise can weaken the training signal, and optimizing a proxy objective does not guarantee improvement in a separate evaluation metric.
3. Experimental design
The experiment defines twelve candidate response styles with four features: helpfulness, correctness, concision, and verbosity. A gold reward combines these features and is reserved for evaluation. A distinct annotator reward generates probabilistic pairwise preferences; noise is injected by flipping a proportion of labels. The three noise levels are 0.05, 0.20, and 0.40. Each condition is repeated across five random seeds. For each seed and noise level, both methods receive the same training preference pairs.
4. Methods
The RLHF baseline fits a linear reward model from pairwise preferences using logistic loss. A categorical policy is then updated with a REINFORCE-style gradient using the learned reward and a KL regularization term. This is a simplified educational approximation, not a full PPO implementation. DPO directly updates categorical policy logits using a pairwise logistic objective relative to a fixed reference policy. Both methods use the same candidate set and matched preference data. Implementation details and hyperparameters are documented in the notebook.
